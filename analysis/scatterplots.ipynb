# Scatterplots of the campaign shapes

The code for these plots was generated by Claude Code.

One u-space ([0,1]^2) scatterplot per copula in `unique_shapes` from
`unique_pairs.py`, each saved as its own PNG in `unique_shapes_scatterplots/`.
No titles, axis labels or ticks — just the points inside the unit square.

`n_samples = 10000`, ten times the main block of `copula_gallery.ipynb`, with the
marker area cut tenfold to match (`s=0.3` vs the gallery's `s=3`, same
`alpha=0.5`). Importing `unique_pairs` is safe: the JSON is only rewritten when
that module runs as a script.

```python
# numpy/torch/matplotlib for this notebook; scipy and statsmodels are imported
# by copula_families (via unique_pairs)
%pip install numpy torch matplotlib scipy statsmodels
```

```python
from pathlib import Path

import numpy as np
import torch
import matplotlib.pyplot as plt

import sys
sys.path.insert(0, str(Path.cwd().parent))  # project root, so the distance_calculation_campaign imports resolve

from distance_calculation_campaign.unique_pairs import unique_shapes
```

```python
n_samples = 10000   # 10x the main block of copula_gallery.ipynb
out_dir = Path("unique_shapes_scatterplots")
out_dir.mkdir(exist_ok=True)

# statsmodels' samplers draw from numpy, the rest from torch -- seed both so
# re-running the notebook reproduces the same pictures
np.random.seed(0)
torch.manual_seed(0)

for name, copula in unique_shapes.items():
    points = np.asarray(copula.sample_u(n_samples))

    fig, ax = plt.subplots(figsize=(2.1, 2.1))
    # marker area cut 10x from the gallery's s=3 to match the 10x point count,
    # so dense regions still show structure instead of a solid blob
    ax.scatter(points[:, 0], points[:, 1], s=0.5, alpha=0.5, linewidths=0)
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_aspect("equal")

    # "(khoudraji_gumbel6_0.4)" -> "khoudraji_gumbel6_0.4.png"
    fig.savefig(out_dir / f"{name.strip('()')}.png", dpi=200, bbox_inches="tight")
    plt.close(fig)

print(f"wrote {len(unique_shapes)} PNGs to {out_dir}/")
```

```python
# The 8 isometries of the square applied to the Khoudraji-Gumbel shape: the
# 4 mirrorings, and the same 4 composed with the transpose. Tags follow
# unique_pairs.py, applied left to right, so "!T!M0" is M0(T(c)) -- the true
# 90-degree rotation. (Mirror and transpose don't commute, but T(M(c)) for all
# M gives the same 8 copulas, just under different names.)
from distance_calculation_campaign.copula_families import MirroredCopula, TransposedCopula

base_name = "(khoudraji_gumbel6_0.4)"
base = unique_shapes[base_name]

ISOMETRIES = {
    "":        lambda c: c,
    "!M0":     lambda c: MirroredCopula(c, 0),
    "!M1":     lambda c: MirroredCopula(c, 1),
    "!M01":    lambda c: MirroredCopula(c, (0, 1)),
    "!T":      lambda c: TransposedCopula(c),
    "!T!M0":   lambda c: MirroredCopula(TransposedCopula(c), 0),
    "!T!M1":   lambda c: MirroredCopula(TransposedCopula(c), 1),
    "!T!M01":  lambda c: MirroredCopula(TransposedCopula(c), (0, 1)),
}

iso_dir = Path("khoudraji_gumbel6_isometries_scatterplots")
iso_dir.mkdir(exist_ok=True)

np.random.seed(0)
torch.manual_seed(0)

for tag, apply in ISOMETRIES.items():
    points = np.asarray(apply(base).sample_u(n_samples))

    fig, ax = plt.subplots(figsize=(2.1, 2.1))
    ax.scatter(points[:, 0], points[:, 1], s=0.5, alpha=0.5, linewidths=0)
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_aspect("equal")

    # e.g. "khoudraji_gumbel6_0.4!T!M0.png"
    fig.savefig(iso_dir / f"{base_name.strip('()')}{tag}.png", dpi=200, bbox_inches="tight")
    plt.close(fig)

print(f"wrote {len(ISOMETRIES)} PNGs to {iso_dir}/")
```
